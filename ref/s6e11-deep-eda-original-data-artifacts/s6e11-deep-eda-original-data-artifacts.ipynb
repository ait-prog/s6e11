## 01 · Imports

```python
from pathlib import Path
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import ks_2samp, chi2_contingency, spearmanr
from scipy.spatial.distance import jensenshannon
from sklearn.metrics import roc_auc_score
from IPython.display import display, Markdown

warnings.filterwarnings('ignore', category=FutureWarning)
pd.set_option('display.max_columns', 60)
pd.set_option('display.max_rows', 70)
pd.set_option('display.float_format', '{:,.4f}'.format)
print('Libraries loaded.')
```

## 02 · Configuration and Visual Style

Here we set the file paths, analysis settings, and chart styling. A consistent palette and shared plotting parameters make the results easier to compare across sections.

```python
SEED = 42
TARGET = 'is_fraud'
ID = 'id'

COMP_DIR = Path('/kaggle/input/competitions/playground-series-s6e11')
ORIGINAL_PATH = Path('/kaggle/input/datasets/uditjain13/credit-card-fraud-detection-2026/credit_card_fraud_2026.csv')

C = {
    'ink': '#17263C', 'muted': '#61758D', 'blue': '#2775D1',
    'red': '#E85A70', 'mint': '#13A99B', 'gold': '#E7A84B',
    'pale': '#F6F9FC', 'grid': '#DDE6EF', 'white': '#FFFFFF'
}

sns.set_theme(style='white', font_scale=0.93)
plt.rcParams.update({
    'figure.facecolor': C['white'], 'axes.facecolor': C['pale'],
    'axes.edgecolor': C['grid'], 'axes.labelcolor': C['ink'],
    'text.color': C['ink'], 'xtick.color': C['muted'],
    'ytick.color': C['muted'], 'axes.titlesize': 11,
    'axes.titleweight': 'semibold', 'font.size': 10,
    'axes.spines.top': False, 'axes.spines.right': False,
    'savefig.bbox': 'tight', 'figure.dpi': 110,
})

def finish(fig, title, note='', top=0.91):
    fig.suptitle(title, x=0.06, y=0.99, ha='left', fontsize=17, fontweight='bold', color=C['ink'])
    if note:
        fig.text(0.06, 0.92 if fig.get_size_inches()[1] <= 8 else 0.945, note, ha='left', fontsize=9, color=C['muted'])
    fig.tight_layout(rect=[0.03, 0.03, 0.98, top])
    plt.show()

print('Visual style configured · consistent colors and typography · no PNG files saved')
```

## 03 · Load Data

We load train, test, and original without modifying them. Transaction IDs have different names in the synthetic and original datasets, so we check feature correspondence rather than identifier names.

```python
train = pd.read_csv(COMP_DIR / 'train.csv')
test = pd.read_csv(COMP_DIR / 'test.csv')
original = pd.read_csv(ORIGINAL_PATH)

FEATURES = [c for c in test.columns if c != ID]
NUM = train[FEATURES].select_dtypes(include='number').columns.tolist()
CAT = train[FEATURES].select_dtypes(include='object').columns.tolist()
BOOL = train[FEATURES].select_dtypes(include='bool').columns.tolist()

assert len(FEATURES) == 24 and len(NUM) == 13 and len(CAT) == 5 and len(BOOL) == 6
assert set(FEATURES).issubset(original.columns)
assert TARGET in train and TARGET in original
assert set(train.columns) == set(test.columns) | {TARGET}

print(f'Train: {train.shape}  |  Test: {test.shape}  |  Original: {original.shape}')
print(f'Features: {len(NUM)} numerical · {len(CAT)} categorical · {len(BOOL)} Boolean')
display(train.head(3))
```

## 06 · Data Quality Checks

We examine missing values, constant columns, unique-value counts, duplicates, and structural consistency across train, test, and original.

```python
audit = pd.DataFrame(index=FEATURES)
audit['Type'] = train[FEATURES].dtypes.astype(str)
audit['Unique in train'] = train[FEATURES].nunique(dropna=False)
audit['Unique in test'] = test[FEATURES].nunique(dropna=False)
audit['Missing in train'] = train[FEATURES].isna().sum()
audit['Missing in test'] = test[FEATURES].isna().sum()
audit['Missing in original'] = original[FEATURES].isna().sum()

display(audit)

print('Full-row duplicates excluding ID:')
print('Train:', train[FEATURES].duplicated().sum())
print('Test:', test[FEATURES].duplicated().sum())
print('Original:', original[FEATURES].duplicated().sum())
print('Unique train IDs:', train[ID].is_unique)
print('Unique test IDs:', test[ID].is_unique)
print('Train/test ID overlap:', len(set(train[ID]) & set(test[ID])))
print('Constant features:', audit.index[audit['Unique in train'] < 2].tolist())
```

## 07 · Target Variable: How Rare Is Fraud?

ROC AUC measures how well transactions are ranked under severe class imbalance, but it does not account for the cost of false positives. We first compare the baseline fraud rate in train and original.

```python
target_counts = pd.DataFrame({
    'Source': ['Train', 'Original'],
    'Transactions': [len(train), len(original)],
    'Fraud': [int(train[TARGET].sum()), int(original[TARGET].sum())]
})
target_counts['Fraud rate, %'] = 100 * target_counts['Fraud'] / target_counts['Transactions']
display(target_counts.style.format({'Fraud rate, %': '{:.3f}%'}).hide(axis='index'))

fig, ax = plt.subplots(figsize=(10, 4))
base = np.arange(2)
rates = target_counts['Fraud rate, %'].values
bars = ax.barh(base, rates, color=[C['blue'], C['gold']], height=.48)
for b, v, n in zip(bars, rates, target_counts['Fraud']):
    ax.text(v + .03, b.get_y() + b.get_height()/2, f'{v:.3f}% · {n:,} cases', va='center', weight='bold')
ax.set_yticks(base, ['Train', 'Original'])
ax.invert_yaxis()
ax.set_xlim(0, max(rates)*1.55)
ax.set_xlabel('Fraud rate, %')
ax.grid(axis='x', alpha=.18)
finish(fig, 'A Rare Event: Around 1.7% of Transactions', 'Labels show the positive-class rate and the absolute number of fraudulent transactions.')
```

## 09 · Numerical Distributions: Train, Test, and Original

For each numerical feature, we plot distributions from all three sources on the same axis. Histograms are density-normalized, so differences in dataset size do not affect bar height.

```python
fig, axes = plt.subplots(4, 4, figsize=(18, 15))
axes = axes.ravel()
plot_rng = np.random.default_rng(SEED)

for ax, c in zip(axes, NUM):
    a = train[c].sample(n=min(45000, len(train)), random_state=SEED).to_numpy()
    b = test[c].sample(n=min(22000, len(test)), random_state=SEED).to_numpy()
    o = original[c].to_numpy()
    lo, hi = np.quantile(np.concatenate([a, b, o]), [.005, .995])
    if hi <= lo:
        hi = lo + 1
    bins = np.linspace(lo, hi, 43)
    ax.hist(a, bins=bins, density=True, histtype='step', linewidth=1.9, color=C['blue'], label='Train')
    ax.hist(b, bins=bins, density=True, histtype='step', linewidth=1.5, color=C['mint'], label='Test')
    ax.hist(o, bins=bins, density=True, histtype='step', linewidth=1.5, color=C['gold'], label='Original')
    ax.set_title(c, loc='left', fontsize=10)
    ax.grid(axis='y', alpha=.13)
for ax in axes[len(NUM):]:
    ax.axis('off')
axes[0].legend(frameon=False, fontsize=8)
finish(fig, '13 Numerical Features · Three Data Sources', 'Train and test are generally similar; original can differ substantially. Identical bins are used for comparison.', top=.92)
```

## 10 · Numerical Feature Signal: Univariate ROC AUC

For each numerical feature, we compute ROC AUC using the raw values, then select the direction that gives an AUC of at least 0.5.

```python
one_auc = pd.DataFrame({
    'Feature': NUM,
    'Raw-value AUC': [roc_auc_score(train[TARGET], train[c]) for c in NUM]
})
one_auc['Best-direction AUC'] = one_auc['Raw-value AUC'].apply(lambda x: max(x, 1-x))
one_auc['Direction'] = np.where(one_auc['Raw-value AUC'] >= .5, '↑', '↓')
one_auc = one_auc.sort_values('Best-direction AUC', ascending=True)
display(one_auc.sort_values('Best-direction AUC', ascending=False).round(5).reset_index(drop=True))

fig, ax = plt.subplots(figsize=(11, 7))
val = one_auc['Best-direction AUC']
ax.barh(one_auc['Feature'], val-.5, left=.5, color=[C['red'] if x >= .70 else C['blue'] for x in val], height=.65)
for i, x in enumerate(val):
    ax.text(x + .004, i, f'{x:.3f}', va='center', fontsize=9)
ax.set_xlim(.49, .82)
ax.axvline(.5, linestyle='--', color=C['muted'], linewidth=1)
ax.set_xlabel('Univariate ROC AUC · direction selected on train')
ax.grid(axis='x', alpha=.18)
finish(fig, 'Where Is the Individual Signal?', 'Selecting direction on train makes this descriptive, not a valid out-of-fold estimate.')
```

## 11 · How Does Fraud Risk Change Across Numerical Features?

We divide each feature into intervals based on train quantiles and calculate the fraud rate within each interval. Group size matters: a result based on a handful of transactions should not be interpreted with the same confidence as one based on hundreds of thousands.

```python
fig, axes = plt.subplots(4, 4, figsize=(18, 15))
axes = axes.ravel()

for ax, c in zip(axes, NUM):
    discrete = train[c].nunique() <= 16
    bins = train[c] if discrete else pd.qcut(train[c], q=10, duplicates='drop')
    stat = train.groupby(bins, observed=True)[TARGET].agg(['mean', 'size'])
    x = np.arange(len(stat))
    ax.plot(x, stat['mean']*100, color=C['red'], marker='o', markersize=4, linewidth=2)
    ax.axhline(train[TARGET].mean()*100, color=C['muted'], linestyle='--', linewidth=1)
    ax.set_title(c, loc='left', fontsize=10)
    ax.set_xlabel('Feature value' if discrete else f'{len(stat)} ascending bins →', fontsize=8)
    ax.set_ylabel('Fraud, %', fontsize=8)
    pos = x[::max(1, len(x)//6)]
    ax.set_xticks(pos, [str(v) if discrete else str(k+1) for k, v in zip(pos, stat.index[::max(1, len(x)//6)])])
    ax.grid(axis='y', alpha=.15)
for ax in axes[len(NUM):]:
    ax.axis('off')
finish(fig, 'How Does Risk Vary Across Numerical Features?', 'Dashed lines indicate the overall train fraud rate. Bin counts depend on the number of unique values.', top=.92)
```

## 12 · Categorical Features: Frequency and Risk

We examine **all five categorical features**. For each category, we compare how often it occurs and what share of its transactions are fraudulent. High risk in a small group does not necessarily mean a large contribution to the total number of fraud cases.

```python
cat_stats = {}
for c in CAT:
    cat_stats[c] = train.groupby(c, dropna=False)[TARGET].agg(['mean', 'size', 'sum']).sort_values('mean', ascending=False)
    cat_stats[c].columns = ['fraud_rate', 'n', 'fraud_n']
    print(f'\n{c}: {DESCR[c]}')
    display(cat_stats[c].assign(fraud_pct=lambda d: d.fraud_rate*100)[['n', 'fraud_n', 'fraud_pct']].round(3))
```

### 13 · Categorical Features: Visual Comparison

After the tables, we look separately at the frequency and fraud rate of each category.

```python
fig, axes = plt.subplots(5, 2, figsize=(17, 23), gridspec_kw={'width_ratios': [1.1, 1.0]})

for row, c in enumerate(CAT):
    st = cat_stats[c].sort_values('n', ascending=True)
    yy = np.arange(len(st))
    ax_l, ax_r = axes[row]
    ax_l.barh(yy, st['n']/len(train)*100, color=C['blue'], height=.65)
    ax_l.set_yticks(yy, st.index.astype(str), fontsize=9)
    ax_l.set_xlabel('Share of transactions, %')
    ax_l.set_title(f'{c} · prevalence', loc='left')
    ax_l.grid(axis='x', alpha=.14)
    ax_r.barh(yy, st['fraud_rate']*100, color=C['red'], height=.65)
    ax_r.axvline(train[TARGET].mean()*100, color=C['muted'], linestyle='--', linewidth=1)
    ax_r.set_yticks(yy, ['']*len(st))
    ax_r.set_xlabel('Fraud rate, %')
    ax_r.set_title('Within-category risk', loc='left')
    ax_r.grid(axis='x', alpha=.14)
finish(fig, 'Categories · Prevalence ≠ Risk', 'Left: share of transactions. Right: fraud rate. Dashed line: overall fraud rate.', top=.956)
```

## 14 · Categorical Risk Map: Frequency vs Intensity

Ranking categories by fraud rate alone can be misleading: a rare category may appear dangerous because of just a few cases. In the first chart, each point represents one value of one of the five categorical features. The X-axis shows transaction count, the Y-axis shows risk relative to the overall rate, and point area corresponds to the number of fraudulent transactions.

The second chart highlights the categories that account for the most **fraudulent transactions in absolute terms**. These are two distinct perspectives: how high the risk is and how much a group contributes to total fraud.

```python
cat_palette = [C['blue'], C['mint'], C['red'], C['gold'], '#8C6BC5']
fraud_base = train[TARGET].mean()
landscape_rows = []
for field in CAT:
    for category, group in cat_stats[field].iterrows():
        n = int(group['n'])
        fraud_n = int(group['fraud_n'])
        risk = float(group['fraud_rate'])
        landscape_rows.append({'feature': field, 'category': str(category), 'n': n,
                               'fraud_n': fraud_n, 'risk': risk, 'lift': risk / fraud_base})
landscape = pd.DataFrame(landscape_rows)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7), gridspec_kw={'width_ratios': [1.17, 1]})
for color, (field, group) in zip(cat_palette, landscape.groupby('feature', sort=False)):
    ax1.scatter(group['n'], group['lift'], s=40 + 10 * np.sqrt(group['fraud_n'].clip(lower=0)),
                c=color, alpha=.78, edgecolors='white', linewidths=.75, label=field, zorder=3)
ax1.axhline(1, color=C['muted'], ls='--', lw=1.5)
ax1.set_xscale('log')
ax1.set_xlabel('Transaction count · logarithmic scale')
ax1.set_ylabel('Fraud lift relative to baseline · ×')
ax1.set_title('All categorical values', loc='left')
ax1.grid(alpha=.15)
ax1.legend(frameon=False, fontsize=8, loc='upper right')
priority = landscape.loc[landscape['n'] >= max(100, int(len(train) * .001))].copy()
priority['label'] = priority['feature'].str.replace('_', ' ', regex=False) + ' · ' + priority['category']
priority = priority.sort_values('fraud_n', ascending=True).tail(11)
colors_by_field = {x: cat_palette[i] for i, x in enumerate(CAT)}
ax2.barh(priority['label'], priority['fraud_n'], color=priority['feature'].map(colors_by_field), height=.65)
for i, row in enumerate(priority.itertuples(index=False)):
    ax2.text(row.fraud_n + max(1, priority['fraud_n'].max()*.01), i,
             f'{row.fraud_n:,} / {row.n:,}  ·  {row.lift:.2f}×', va='center', fontsize=8)
ax2.set_xlim(0, priority['fraud_n'].max()*1.65 if len(priority) else 1)
ax2.set_title('Categories with the Most Fraud Cases', loc='left')
ax2.set_xlabel('Number of fraudulent transactions')
ax2.grid(axis='x', alpha=.15)
finish(fig, 'Fraud Risk Landscape · How Widespread Is the Risk?',
       'Each point is a category; size ≈ number of fraud cases. Compare risk with its supporting sample size.', top=.85)
display(landscape.sort_values('fraud_n', ascending=False).head(12).round(4).reset_index(drop=True))
```

## 15 · Fraud in Dollars: Frequency and Monetary Volume

Counting fraudulent transactions reveals where fraud occurs most often, but not where the largest transaction amounts are concentrated. Here **fraud amount** means the sum of `amount_usd` for transactions labeled `is_fraud = 1`. It is a descriptive measure of transaction value, **not confirmed financial loss**.

In synthetic train, fraudulent transactions make up **1.67% of operations** but account for **2.31% of total transaction value** ($1.29 million out of $55.92 million). Electronics contributes **46.9% of fraud-tagged dollars**, compared with **12.8% of fraud cases**; Travel contributes **21.4% of dollars** versus **4.6% of cases**. Gift Cards has the highest category fraud rate (**7.30%**), yet accounts for only **2.0% of fraud-tagged dollars**.

The comparison with original offers a useful contrast: its overall fraud-tagged value share is also about **2.32%**, but its category-level allocation differs substantially. Original has only **339 fraud cases**, so its category breakdown is less stable. The two charts distinguish fraud **frequency** from fraud **value** without assuming that either measure alone describes operational impact.

```python
# Transaction values tagged as fraud, not verified monetary losses.
fraud_money_sources = {}
for source_name, data in [('Synthetic train', train), ('Original', original)]:
    working = data.assign(fraud_amount_usd=np.where(data[TARGET].eq(1), data['amount_usd'], 0.0))
    detail = working.groupby('merchant_category', observed=True).agg(
        transactions=(TARGET, 'size'),
        fraud_cases=(TARGET, 'sum'),
        transaction_amount_usd=('amount_usd', 'sum'),
        fraud_amount_usd=('fraud_amount_usd', 'sum')
    )
    detail['fraud_rate_pct'] = 100 * detail['fraud_cases'] / detail['transactions']
    detail['share_fraud_cases_pct'] = 100 * detail['fraud_cases'] / detail['fraud_cases'].sum()
    detail['share_fraud_amount_pct'] = 100 * detail['fraud_amount_usd'] / detail['fraud_amount_usd'].sum()
    fraud_money_sources[source_name] = detail

    total_amount = data['amount_usd'].sum()
    tagged_amount = detail['fraud_amount_usd'].sum()
    print(f'{source_name}: {int(data[TARGET].sum()):,} fraud cases / {len(data):,} transactions '
          f'({100*data[TARGET].mean():.2f}%); '
          f'${tagged_amount:,.2f} tagged as fraud / ${total_amount:,.2f} total '
          f'({100*tagged_amount/total_amount:.2f}%).')

money_train = fraud_money_sources['Synthetic train'].sort_values('fraud_amount_usd', ascending=False)
money_orig = fraud_money_sources['Original'].reindex(money_train.index)
display(money_train[[
    'transactions', 'fraud_cases', 'fraud_rate_pct', 'fraud_amount_usd',
    'share_fraud_cases_pct', 'share_fraud_amount_pct'
]].round(2))

plot_categories = money_train.index[::-1]
y = np.arange(len(plot_categories))
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(19, 8.6), gridspec_kw={'width_ratios': [1.05, 1]})
ax1.barh(y - .2, money_train.loc[plot_categories, 'share_fraud_cases_pct'],
         height=.38, color=C['blue'], label='Share of fraud cases')
ax1.barh(y + .2, money_train.loc[plot_categories, 'share_fraud_amount_pct'],
         height=.38, color=C['red'], label='Share of fraud-tagged dollars')
ax1.set_yticks(y, plot_categories)
ax1.set_xlabel('Share of synthetic-train fraud, %')
ax1.set_title('Fraud Frequency vs Dollar Exposure', loc='left')
ax1.legend(frameon=False, fontsize=8)
ax1.grid(axis='x', alpha=.15)

ax2.barh(y - .2, money_train.loc[plot_categories, 'share_fraud_amount_pct'],
         height=.38, color=C['blue'], label='Synthetic train')
ax2.barh(y + .2, money_orig.loc[plot_categories, 'share_fraud_amount_pct'],
         height=.38, color=C['gold'], label='Original')
ax2.set_yticks(y, plot_categories)
ax2.set_xlabel('Share of fraud-tagged dollars, %')
ax2.set_title('How the Dollar Mix Changes', loc='left')
ax2.legend(frameon=False, fontsize=8)
ax2.grid(axis='x', alpha=.15)
finish(fig, 'Fraud in Dollars · Category-Level Exposure',
       'Dollar exposure = value of transactions marked as fraud; original category shares are based on 339 cases.', top=.90)
```

## 16 · Where Fraud-Tagged Transaction Value Concentrates

Large transactions may account for relatively few fraud cases but a disproportionately large share of the total value of transactions labeled as fraud. We divide each dataset into ten equally sized groups by `amount_usd` and compare three measures: share of total transaction value, share of fraud cases, and share of fraud-tagged transaction value. The top-decile threshold is calculated separately for train, test, and original.


```python
# The largest-amount decile is defined independently within each dataset.
# Monetary exposure means the amount of transactions labeled as fraud, not verified bank losses.
tail_records = []
tail_details = {}
for source_name, frame in [('Train', train), ('Test', test), ('Original', original)]:
    threshold = frame['amount_usd'].quantile(0.9)
    top = frame['amount_usd'].ge(threshold)
    row = {
        'Датасет': source_name,
        'Порог верхних 10%, $': threshold,
        'Доля оборота в верхних 10%, %': 100 * frame.loc[top, 'amount_usd'].sum() / frame['amount_usd'].sum(),
    }
    if TARGET in frame.columns:
        fraud = frame[TARGET].eq(1)
        fraud_dollars = frame['amount_usd'].where(fraud, 0.0)
        row['Доля fraud-случаев в верхних 10%, %'] = 100 * (fraud & top).sum() / fraud.sum()
        row['Доля fraud-суммы в верхних 10%, %'] = 100 * fraud_dollars[top].sum() / fraud_dollars.sum()
        decile = pd.qcut(frame['amount_usd'].rank(method='first'), 10, labels=False) + 1
        tail_details[source_name] = (fraud_dollars.groupby(decile).sum() / fraud_dollars.sum() * 100).reindex(range(1, 11))
    tail_records.append(row)

tail_comparison = pd.DataFrame(tail_records).set_index('Датасет')
display(tail_comparison.round(2))

fig, axes = plt.subplots(1, 2, figsize=(18, 6.5), gridspec_kw={'width_ratios': [1.1, 1]})
metrics = ['Доля оборота в верхних 10%, %', 'Доля fraud-случаев в верхних 10%, %', 'Доля fraud-суммы в верхних 10%, %']
metric_labels = ['Все суммы', 'Случаи fraud', 'Сумма fraud']
x = np.arange(len(metrics))
for label, offset, color in [('Train', -0.18, C['blue']), ('Original', 0.18, C['gold'])]:
    values = tail_comparison.loc[label, metrics].to_numpy(dtype=float)
    bars = axes[0].bar(x + offset, values, width=0.34, color=color, label=label)
    axes[0].bar_label(bars, labels=[f'{v:.1f}%' for v in values], padding=3, fontsize=9)
axes[0].set_xticks(x, metric_labels)
axes[0].set_ylim(0, 94)
axes[0].set_ylabel('Доля показателя в верхних 10% операций, %')
axes[0].set_title('Что находится в верхнем дециле', loc='left')
axes[0].legend(frameon=False)
axes[0].grid(axis='y', alpha=0.17)

deciles = np.arange(1, 11)
for label, color in [('Train', C['blue']), ('Original', C['gold'])]:
    axes[1].plot(deciles, tail_details[label], color=color, marker='o', lw=2.2, label=label)
axes[1].axvspan(9.6, 10.4, color=C['red'], alpha=0.08)
axes[1].set_xticks(deciles)
axes[1].set_xlabel('Дециль суммы транзакции (10 = крупнейшие)')
axes[1].set_ylabel('Доля денежного объёма fraud, %')
axes[1].set_title('Концентрация fraud-суммы по децилям', loc='left')
axes[1].grid(axis='y', alpha=0.17)
axes[1].legend(frameon=False)
finish(fig, 'Fraud в денежном выражении · верхние 10% транзакций',
       'Группы построены по amount_usd внутри каждого датасета; суммы не равны подтверждённым потерям.', top=0.84)
```

## 18 · Context Matters: Fraud Flags Across Transaction Channels

`billing_shipping_mismatch` is much more common in the Online channel: **2.81% of transactions**, compared with **0.32%** in other channels. Yet among transactions with this flag, the fraud rate is *lower* in Online (**12.48% vs 15.87%**). Stratifying by `merchant_category` and `auth_method` preserves the direction of the difference (pooled odds ratio approximately **0.73**).

The chart on the left shows how often the flag fires; the one on the right shows fraud risk among flagged transactions. A second example, `is_ai_generated_scam_attempt`, offers a comparison: positive flags outside Online/In-App are rare, so the risk estimate there is considerably less precise. We use 95% Wilson intervals for small groups. These are differences between observed groups, not proof that the channel itself causes fraud.

```python
from statsmodels.stats.proportion import proportion_confint
from statsmodels.stats.contingency_tables import StratifiedTable

# Compare signal prevalence (train/test) and labeled risk (train only) by channel.
context_flags = ['billing_shipping_mismatch', 'is_ai_generated_scam_attempt']
context_channels = ['Online', 'In-App', 'POS', 'Contactless', 'ATM']
context_rows = []
for flag in context_flags:
    for channel in context_channels:
        train_slice = train.loc[train['channel'].eq(channel)]
        test_slice = test.loc[test['channel'].eq(channel)]
        active = train_slice.loc[train_slice[flag].eq(True), TARGET]
        count, positive = len(active), int(active.sum())
        low, high = proportion_confint(positive, count, method='wilson') if count else (np.nan, np.nan)
        context_rows.append({
            'Flag': flag, 'Channel': channel,
            'Train prevalence, %': 100 * train_slice[flag].mean(),
            'Test prevalence, %': 100 * test_slice[flag].mean(),
            'Active train rows': count, 'Fraud among active': positive,
            'Fraud rate among active, %': 100 * positive / count if count else np.nan,
            'CI low, %': 100 * low, 'CI high, %': 100 * high,
        })
context_table = pd.DataFrame(context_rows)

# Control for merchant category and authentication method within flagged transactions.
strata = []
flagged_billing = train.loc[train['billing_shipping_mismatch'].eq(True)]
for _, group in flagged_billing.groupby(['merchant_category', 'auth_method'], observed=True):
    contingency = pd.crosstab(group['channel'].eq('Online'), group[TARGET]).reindex(
        index=[False, True], columns=[0, 1], fill_value=0).to_numpy()
    if contingency.sum(axis=1).min() > 0:
        strata.append(contingency)
cmh = StratifiedTable(strata)
print('Billing/shipping mismatch · Online vs other channels, adjusted for category × auth:')
print(f'  pooled OR = {cmh.oddsratio_pooled:.3f}; 95% CI = '
      f'[{cmh.oddsratio_pooled_confint()[0]:.3f}, {cmh.oddsratio_pooled_confint()[1]:.3f}]; '
      f'p = {cmh.test_null_odds().pvalue:.4g}; strata = {len(strata)}')
display(context_table[['Flag', 'Channel', 'Active train rows',
                       'Train prevalence, %', 'Test prevalence, %',
                       'Fraud rate among active, %', 'CI low, %', 'CI high, %']].round(3))

fig, axes = plt.subplots(2, 2, figsize=(16.5, 10.8))
xs = np.arange(len(context_channels))
for row, flag in enumerate(context_flags):
    subset = context_table[context_table['Flag'].eq(flag)].set_index('Channel').reindex(context_channels)
    ax_prev, ax_risk = axes[row]
    ax_prev.bar(xs, subset['Train prevalence, %'], width=.58,
                color=C['blue'] if row == 0 else C['mint'], alpha=.92, label='Train')
    ax_prev.scatter(xs, subset['Test prevalence, %'], marker='D', s=37,
                    color=C['gold'], edgecolor='white', linewidth=.8, zorder=4, label='Test')
    ax_prev.set_title(('Billing / shipping mismatch' if row == 0 else 'AI scam-attempt flag') + ' · Prevalence', loc='left')
    ax_prev.set_ylabel('Flag prevalence, %')
    ax_prev.set_xticks(xs, context_channels)
    ax_prev.grid(axis='y', alpha=.16)
    if row == 0: ax_prev.legend(frameon=False, loc='upper right', fontsize=9)

    p = subset['Fraud rate among active, %'].to_numpy()
    err = np.vstack([p - subset['CI low, %'].to_numpy(), subset['CI high, %'].to_numpy() - p])
    ax_risk.bar(xs, p, width=.58, color=C['red'] if row == 0 else C['gold'], alpha=.88)
    ax_risk.errorbar(xs, p, yerr=err, fmt='none', ecolor=C['ink'], capsize=4, lw=1.3)
    for j, val in enumerate(subset['Active train rows']):
        ax_risk.text(j, max(0, p[j]-1), f'n={int(val):,}', ha='center', va='top',
                     fontsize=8, fontweight='semibold', color='#17263C')
    ax_risk.set_xticks(xs, context_channels)
    ax_risk.set_ylabel('Fraud rate when flagged, %')
    ax_risk.set_ylim(0, max(25, float(np.nanmax(subset['CI high, %'])) * 1.17))
    ax_risk.set_title(('Billing / shipping mismatch' if row == 0 else 'AI scam-attempt flag') + ' · Conditional risk', loc='left')
    ax_risk.grid(axis='y', alpha=.16)
finish(fig, 'The Same Flag Can Mean Different Things Across Channels',
       'Diamonds = test prevalence · whiskers = Wilson 95% CI · rare flags have wide intervals.', top=.90)
```

## 21 · Risk Concentration and Fraud Coverage

A fraud-rate chart alone does not show how much fraud is found in each group. Here we calculate both **risk within the group** and **the share of all fraudulent transactions** attributable to it, for synthetic train and original.

In train, transactions with no active flags account for approximately **33.1% of all fraud**. The group with three flags is much riskier per transaction, but accounts for only about **5.6% of all fraud**. We combine four or more flags for very small groups and display sample sizes.

```python
from statsmodels.stats.proportion import proportion_confint

# Aggregate the six Boolean indicators without altering either dataset.
coverage_records = []
for source_name, df in [('Synthetic train', train), ('Original', original)]:
    flag_count = df[BOOL].sum(axis=1).clip(upper=4).astype(int)
    st = df.groupby(flag_count)[TARGET].agg(['size', 'sum']).reindex(range(5), fill_value=0)
    for flags, row in st.iterrows():
        n, positives = int(row['size']), int(row['sum'])
        lo, hi = proportion_confint(positives, n, method='wilson') if n else (np.nan, np.nan)
        coverage_records.append({'Источник': source_name, 'Флагов': '4+' if flags == 4 else str(flags),
                                 'Операций': n, 'Fraud': positives,
                                 'Fraud rate, %': 100 * positives / n if n else np.nan,
                                 'Доля всего fraud, %': 100 * positives / df[TARGET].sum(),
                                 'Нижняя 95%, %': 100 * lo, 'Верхняя 95%, %': 100 * hi})
coverage_v9 = pd.DataFrame(coverage_records)
fig, axes = plt.subplots(1, 2, figsize=(17, 6))
labels = ['0', '1', '2', '3', '4+']
xx = np.arange(len(labels))
for src, dx, color in [('Synthetic train', -.18, C['blue']), ('Original', .18, C['gold'])]:
    sub = coverage_v9[coverage_v9['Источник'] == src].set_index('Флагов').reindex(labels)
    axes[0].bar(xx+dx, sub['Fraud rate, %'], width=.35, color=color, label=src)
    axes[1].bar(xx+dx, sub['Доля всего fraud, %'], width=.35, color=color, label=src)
for ax, title, ylabel in zip(axes, ['Риск внутри группы', 'Доля всех мошеннических операций'],
                              ['Fraud rate, %', 'Доля fraud, %']):
    ax.set_xticks(xx, labels)
    ax.set_xlabel('Количество активных Boolean-флагов (4+ объединены)')
    ax.set_ylabel(ylabel)
    ax.set_title(title, loc='left')
    ax.grid(axis='y', alpha=.15)
axes[0].legend(frameon=False)
finish(fig, 'Высокий риск ≠ большой охват мошенничества',
       'Сравнение synthetic train и original; в редких группах уровень риска оценивается неточно.', top=.85)
display(coverage_v9.round(3).reset_index(drop=True))
```

## 22 · The Monetary Blind Spot of Fraud Warning Flags

We previously found fraudulent transactions with none of the Boolean flags active. Now we also separate them by transaction size: below or above the 90th percentile of `amount_usd` within each dataset. ## 27 · Interaction Atlas: Risk Beyond a Baseline Expectation

A high fraud rate for a combination of two features does not necessarily indicate an interaction. We compare observed risk with **a baseline that assumes no pairwise interaction**: expected risk is derived from the log-odds of the individual features and the overall fraud rate. The heatmaps show the ratio of *observed* to *expected* risk.

**1.00×** indicates agreement with the baseline; **above 1.00×** indicates additional risk relative to that baseline. Each cell displays its group size; groups that are too small are hidden. This is a descriptive check, not a formal test of interaction.

```python
# Additive-log-odds reference: logit(p_ab) ~= logit(p_a) + logit(p_b) - logit(p_all).
def _smooth_p(pos, total):
    return (pos + .5) / (total + 1.)

def _logit(p):
    p = np.clip(p, 1e-8, 1 - 1e-8)
    return np.log(p / (1 - p))

def _atlas_matrix(df, row, col, min_count, min_fraud):
    data = pd.DataFrame({'row': row, 'col': col, 'y': df[TARGET].astype('int8')}).dropna(subset=['row', 'col'])
    rows = sorted(data['row'].unique())
    cols = sorted(data['col'].unique())
    base = _smooth_p(data['y'].sum(), len(data))
    rstats = data.groupby('row', observed=True)['y'].agg(['sum','size'])
    cstats = data.groupby('col', observed=True)['y'].agg(['sum','size'])
    joint = data.groupby(['row','col'], observed=True)['y'].agg(['sum','size'])
    lift = pd.DataFrame(np.nan, index=rows, columns=cols)
    annotations = pd.DataFrame('', index=rows, columns=cols)
    for (r,c), rec in joint.iterrows():
        n, pos = int(rec['size']), int(rec['sum'])
        if n < min_count or pos < min_fraud:
            annotations.loc[r,c] = f'n={n:,}\ninsufficient data'
            continue
        pr = _smooth_p(rstats.loc[r,'sum'], rstats.loc[r,'size'])
        pc = _smooth_p(cstats.loc[c,'sum'], cstats.loc[c,'size'])
        expected = 1 / (1 + np.exp(-(_logit(pr) + _logit(pc) - _logit(base))))
        observed = _smooth_p(pos, n)
        ratio = observed / expected
        lift.loc[r,c] = np.log2(max(ratio, 1e-8))
        annotations.loc[r,c] = f'{ratio:.2f}×\nn={n:,}'
    return lift.astype(float), annotations

quantile = lambda x: pd.qcut(x, q=5, labels=False, duplicates='drop').map(lambda v: f'Q{int(v)+1}' if pd.notna(v) else np.nan)
cvv_bucket = train['cvv_retry_count'].clip(upper=3).map(lambda v: '3+' if v >= 3 else str(int(v)))
pairs_atlas = [
    ('Merchant risk × CVV retries', quantile(train['merchant_risk_score']), cvv_bucket),
    ('Velocity × hours since last txn', quantile(train['velocity_score']), quantile(train['hours_since_last_txn'])),
    ('VPN × IP country mismatch', train['used_vpn'].map({False:'No', True:'Yes'}),
     train['ip_country_mismatch'].map({False:'No', True:'Yes'})),
]
min_n = min(200, max(5, len(train) // 1500))
min_frauds = 5 if int(train[TARGET].sum()) > 200 else 1
fig, axes = plt.subplots(1, 3, figsize=(19, 6.2), gridspec_kw={'width_ratios':[1.08,1.08,.84]})
for i, (ax, (name, rows, cols)) in enumerate(zip(axes, pairs_atlas)):
    values, notes = _atlas_matrix(train, rows, cols, min_n, min_frauds)
    sns.heatmap(values, ax=ax, cmap='RdBu_r', vmin=-1.3, vmax=1.3, center=0,
                linewidths=1, linecolor=C['white'], square=False, annot=notes.to_numpy(), fmt='',
                cbar=i==2, cbar_kws={'label':'log₂(observed / expected)', 'shrink':.65},
                annot_kws={'size': 8})
    ax.set_title(name, loc='left', fontsize=10, fontweight='bold')
    ax.set_xlabel('Feature B')
    ax.set_ylabel('Feature A' if i==0 else '')
    ax.tick_params(axis='x', rotation=0)
    ax.tick_params(axis='y', rotation=0)
finish(fig, 'Fraud Risk Synergy Atlas · Where Does Risk Exceed Expectations?',
       f'Reference: additive log-odds; cells with n < {min_n} or fraud < {min_frauds} excluded.', top=.83)
```

## 28 · Distribution Shift Between Train and Test

We assess how closely input-feature distributions match between train and test. For numerical features, we use the Kolmogorov–Smirnov statistic (KS); for categorical and Boolean features, we use Jensen–Shannon distance.

```python
drift_rows = []
for c in NUM:
    ks = ks_2samp(train[c], test[c], method='asymp').statistic
    drift_rows.append([c, 'numeric', 'KS', ks])
for c in CAT + BOOL:
    a = train[c].value_counts(normalize=True)
    b = test[c].value_counts(normalize=True)
    idx = a.index.union(b.index)
    js = jensenshannon(a.reindex(idx, fill_value=0), b.reindex(idx, fill_value=0), base=2)
    drift_rows.append([c, 'category/boolean', 'JS distance', js])
drift = pd.DataFrame(drift_rows, columns=['Feature', 'Type', 'Metric', 'Value'])
display(drift.sort_values(['Type', 'Value'], ascending=[True, False]).round(5).reset_index(drop=True))

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
for ax, part, kind, color in [(axes[0], drift[drift['Metric']=='KS'], 'Numerical · KS', C['blue']),
                               (axes[1], drift[drift['Metric']=='JS distance'], 'Categories / flags · JS', C['mint'])]:
    s = part.sort_values('Value')
    ax.barh(s['Feature'], s['Value'], color=color, height=.7)
    ax.set_title(kind, loc='left')
    ax.set_xlabel('Distance between train and test')
    ax.grid(axis='x', alpha=.18)
finish(fig, 'Train/Test Distribution Shift', 'KS and JS have different meanings and scales; their numerical values should not be compared across panels.', top=.88)
```

## 29 · Joint Train/Test Shift: Feature Pairs

Similar marginal distributions do not guarantee similar joint distributions. We examine eight pairs of categorical and Boolean features using **JS distance between their full joint frequency tables**.

Initial calculations show very small train/test distances for these pairs (approximately **0.002–0.008**), while train/original distances are much larger. This metric also reflects changes in marginal frequencies; it does not isolate the strength of association between the two features.

```python
from scipy.spatial.distance import jensenshannon

joint_pairs = [
    ('merchant_category','auth_method'), ('merchant_category','channel'),
    ('channel','device_type'), ('auth_method','cvv_retry_count'),
    ('merchant_category','is_new_merchant'),
    ('merchant_category','is_ai_generated_scam_attempt'),
    ('is_foreign_transaction','ip_country_mismatch'), ('is_new_merchant','used_vpn')]

def pair_js(df_a, df_b, col_a, col_b):
    # Align every combination before comparing joint probability distributions.
    p = df_a.groupby([col_a,col_b],dropna=False).size()
    q = df_b.groupby([col_a,col_b],dropna=False).size()
    universe = p.index.union(q.index)
    p = p.reindex(universe,fill_value=0).to_numpy(dtype=float)
    q = q.reindex(universe,fill_value=0).to_numpy(dtype=float)
    return float(jensenshannon(p/p.sum(),q/q.sum(),base=2))

joint_rows = []
for a,b in joint_pairs:
    joint_rows.append({'Пара':f'{a} × {b}',
                       'Train/Test JS':pair_js(train,test,a,b),
                       'Train/Original JS':pair_js(train,original,a,b)})
joint_v9 = pd.DataFrame(joint_rows).sort_values('Train/Original JS')
fig, ax = plt.subplots(figsize=(14.3,7.0))
yy=np.arange(len(joint_v9))
ax.barh(yy-.19,joint_v9['Train/Test JS'],height=.35,color=C['blue'],label='Train/Test')
ax.barh(yy+.19,joint_v9['Train/Original JS'],height=.35,color=C['gold'],label='Train/Original')
ax.set_yticks(yy,joint_v9['Пара'],fontsize=8)
ax.set_xlabel('Jensen–Shannon distance, base 2')
ax.legend(frameon=False)
ax.grid(axis='x',alpha=.15)
finish(fig,'Сдвиг отдельных признаков и их сочетаний — разные вопросы',
       'Сравниваются полные совместные распределения выбранных пар, без target.',top=.86)
display(joint_v9.sort_values('Train/Test JS',ascending=False).round(5).reset_index(drop=True))
```

## 31 · How Have Numerical Feature Dependencies Changed?

Feature distributions can appear similar even when relationships between features differ. We compare Spearman correlations in train, test, and original across **all 78 pairs of the 13 numerical features**. The largest gap concerns `hours_since_last_txn × velocity_score`: approximately **−0.445 in synthetic** versus **−0.355 in original**.

We check whether the effect repeats in test and estimate the uncertainty of the difference using bootstrap resampling of equally sized 20,000-row samples. We also examine how median `velocity_score` varies across transaction-time intervals. A change in correlation indicates a change in statistical dependence; it does not, by itself, identify the cause or imply that the generator deliberately altered this pair.

```python
from itertools import combinations

# Compare each pair on the same rank-correlation definition.
dep_corr = {name: df[NUM].corr(method='spearman') for name, df in [
    ('Train', train), ('Test', test), ('Original', original)]}
dep_rows = []
for a, b in combinations(NUM, 2):
    r_tr, r_te, r_or = [dep_corr[name].loc[a, b] for name in ('Train', 'Test', 'Original')]
    dep_rows.append({'Pair': f'{a} × {b}', 'Train rho': r_tr,
                     'Test rho': r_te, 'Original rho': r_or,
                     'Delta Train-Original': r_tr - r_or,
                     '|Delta|': abs(r_tr-r_or)})
dep_table = pd.DataFrame(dep_rows).sort_values('|Delta|', ascending=False).reset_index(drop=True)
print(f'Numeric pairs compared: {len(dep_table)}')
display(dep_table.head(10).round(4).drop(columns='|Delta|'))

# Bootstrap two equal-size samples for the largest observed gap; other pairs are exploratory.
rng_dep = np.random.default_rng(SEED)
cols_dep = ['hours_since_last_txn', 'velocity_score']
train_dep = train[cols_dep].sample(n=20_000, random_state=SEED).to_numpy()
original_dep = original[cols_dep].to_numpy()
gap_samples = []
for _ in range(180):
    i_tr = rng_dep.integers(0, len(train_dep), size=len(train_dep))
    i_or = rng_dep.integers(0, len(original_dep), size=len(original_dep))
    tr_rho = spearmanr(train_dep[i_tr, 0], train_dep[i_tr, 1]).statistic
    or_rho = spearmanr(original_dep[i_or, 0], original_dep[i_or, 1]).statistic
    gap_samples.append(tr_rho - or_rho)
gap_ci = np.quantile(gap_samples, [0.025, 0.975])
print(f'Largest gap (Train - Original): {dep_table.loc[0, "Delta Train-Original"]:+.4f}')
print(f'Bootstrap 95% interval on 20k/20k matched samples: [{gap_ci[0]:+.4f}, {gap_ci[1]:+.4f}]')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(17.5, 7.2), gridspec_kw={'width_ratios': [1.1, 1]})
top = dep_table.head(7).iloc[::-1]
y = np.arange(len(top))
ax1.scatter(top['Original rho'], y, s=65, color=C['gold'], zorder=3, label='Original')
ax1.scatter(top['Train rho'], y, s=65, color=C['blue'], zorder=3, label='Synthetic train')
for j, (_, row) in enumerate(top.iterrows()):
    ax1.plot([row['Original rho'], row['Train rho']], [j, j],
             color='#AFC0D2', lw=2.5, zorder=1)
ax1.set_yticks(y, top['Pair'].str.replace(' × ', '\n× ', regex=False))
ax1.tick_params(axis='y', labelsize=8)
ax1.axvline(0, color=C['grid'], lw=1)
ax1.set_xlabel('Spearman rho')
ax1.set_title('Largest pairwise dependency shifts', loc='left')
ax1.legend(frameon=False, fontsize=9)
ax1.grid(axis='x', alpha=.13)

# Use the original time quantiles as fixed bins for all three datasets.
quantiles = np.quantile(original['hours_since_last_txn'], np.linspace(.1, .9, 9))
edges = np.r_[-np.inf, np.unique(quantiles), np.inf]
for label, frame, color, marker in [('Original', original, C['gold'], 'o'),
                                     ('Synthetic train', train, C['blue'], 's'),
                                     ('Synthetic test', test, C['mint'], '^')]:
    bucket = pd.cut(frame['hours_since_last_txn'], bins=edges)
    median = frame.groupby(bucket, observed=False)['velocity_score'].median().to_numpy()
    ax2.plot(np.arange(len(median)), median, color=color, marker=marker,
             lw=2.1, markersize=5.5, label=label)
ax2.set_xticks(np.arange(len(edges)-1), [str(i+1) for i in range(len(edges)-1)])
ax2.set_xlabel('Time-since-last-transaction bin (original deciles)')
ax2.set_ylabel('Median velocity_score')
ax2.set_title('Risk-speed profile across fixed time bins', loc='left')
ax2.grid(axis='y', alpha=.15)
ax2.legend(frameon=False, fontsize=9)
finish(fig, 'Synthetic Dependencies: Similar Variables, Different Relationships',
       '78 Spearman pairs; displayed gaps are exploratory. The time bins are fixed using original quantiles.', top=.87)
```

## 33 · How Reliable Are Differences in Univariate AUC?

When comparing synthetic train and original, we cannot independently choose whichever AUC direction is higher in each dataset. Here, **the direction is fixed using train** and then applied unchanged to original. For original, we show approximate 95% intervals from 150 stratified bootstrap resamples.


```python
orig_y = original[TARGET].astype(int).to_numpy()
orig_pos = np.flatnonzero(orig_y == 1)
orig_neg = np.flatnonzero(orig_y == 0)
rng_auc = np.random.default_rng(SEED + 18)
boot_repeats = 150
# The same class-stratified bootstrap draws are reused across all features.
boot_idx = [np.concatenate([rng_auc.choice(orig_pos, len(orig_pos), replace=True),
                            rng_auc.choice(orig_neg, len(orig_neg), replace=True)])
            for _ in range(boot_repeats)]
auc_ci = []
for feature in NUM:
    train_auc_raw = roc_auc_score(train[TARGET], train[feature])
    direction = 1 if train_auc_raw >= .5 else -1
    raw_original = original[feature].to_numpy(dtype=float) * direction
    point_original = roc_auc_score(orig_y, raw_original)
    boot = [roc_auc_score(orig_y[idx], raw_original[idx]) for idx in boot_idx]
    lo, hi = np.quantile(boot, [.025,.975])
    auc_ci.append({'feature':feature, 'train_auc':max(train_auc_raw,1-train_auc_raw),
                   'original_auc_fixed':point_original, 'low':lo, 'high':hi,
                   'difference':max(train_auc_raw,1-train_auc_raw)-point_original})
auc_ci = pd.DataFrame(auc_ci).sort_values('difference', key=abs).reset_index(drop=True)
fig, ax = plt.subplots(figsize=(12.5, 7.2))
yy = np.arange(len(auc_ci))
ax.hlines(yy, auc_ci['train_auc'], auc_ci['original_auc_fixed'], color=C['grid'], linewidth=2, zorder=1)
ax.errorbar(auc_ci['original_auc_fixed'], yy,
            xerr=np.vstack([auc_ci['original_auc_fixed']-auc_ci['low'], auc_ci['high']-auc_ci['original_auc_fixed']]),
            fmt='o', color=C['gold'], elinewidth=1.6, capsize=3, ms=6,
            label='Original · 95% bootstrap CI', zorder=3)
ax.scatter(auc_ci['train_auc'], yy, c=C['blue'], s=46, label='Synthetic train', zorder=4)
ax.axvline(.5, color=C['muted'], lw=1, ls='--')
ax.set_yticks(yy, auc_ci['feature'])
ax.invert_yaxis()
ax.set_xlabel('ROC AUC · feature direction fixed on train')
ax.set_xlim(max(.36, min(auc_ci['low'].min(),.48)-.02), min(.92,max(auc_ci['high'].max(),auc_ci['train_auc'].max())+.06))
ax.grid(axis='x', alpha=.13)
ax.legend(loc='lower right', frameon=False)
finish(fig, 'Univariate Signal · Differences and Uncertainty',
       f'Original contains {len(orig_pos):,} fraud and {len(orig_neg):,} non-fraud rows; 150 stratified bootstrap resamples.', top=.88)
display(auc_ci.round(4).rename(columns={'feature':'Feature','train_auc':'Train AUC',
    'original_auc_fixed':'Original AUC', 'low':'CI low', 'high':'CI high', 'difference':'Δ AUC'}))
```

## 34 · The Disappearing Time Signal: Transaction Hour

In original, the fraud rate between 00:00 and 05:59 is **2.76%**, noticeably higher than during the day. In synthetic train, nighttime transactions have a fraud rate of around **1.66%**, almost the same as transactions at other times.

Original is too small for a smooth hour-by-hour curve, so we focus our interpretation on **four six-hour intervals**. We show Wilson 95% intervals, group sizes, and a separate χ² test of association with time of day.

```python
from scipy.stats import chi2_contingency
from statsmodels.stats.proportion import proportion_confint

time_blocks = [(0,5,'00–06'),(6,11,'06–12'),(12,17,'12–18'),(18,23,'18–24')]
time_rows=[]
for name,df in [('Synthetic train',train),('Original',original)]:
    for low,high,label in time_blocks:
        y=df.loc[df['time_of_day_hour'].between(low,high),TARGET]
        n,k=len(y),int(y.sum())
        lo,hi=proportion_confint(k,n,method='wilson')
        time_rows.append({'Источник':name,'Время':label,'Операций':n,'Fraud':k,
                          'Fraud rate, %':100*k/n,'CI low, %':100*lo,'CI high, %':100*hi})
time_v9=pd.DataFrame(time_rows)
fig,axes=plt.subplots(1,2,figsize=(17,6))
xx=np.arange(4)
for name,color,dx in [('Synthetic train',C['blue'],-.18),('Original',C['gold'],.18)]:
    sub=time_v9[time_v9['Источник']==name]
    axes[0].bar(xx+dx,sub['Fraud rate, %'],width=.35,color=color,label=name)
    means=sub['Fraud rate, %'].to_numpy()
    errors=np.array([means-sub['CI low, %'].to_numpy(), sub['CI high, %'].to_numpy()-means])
    axes[0].errorbar(xx+dx,means,yerr=errors,fmt='none',ecolor=C['ink'],capsize=3,lw=1)
    axes[1].plot(xx,means,marker='o',lw=2,color=color,label=name)
axes[0].set_xticks(xx,[x[2] for x in time_blocks]);axes[1].set_xticks(xx,[x[2] for x in time_blocks])
axes[0].set_title('Риск и 95% интервалы',loc='left')
axes[1].set_title('Форма временного профиля',loc='left')
for ax in axes:
    ax.set_ylabel('Fraud rate, %')
    ax.set_xlabel('Время суток')
    ax.grid(axis='y',alpha=.15)
axes[0].legend(frameon=False)
finish(fig,'Ночной риск в original и почти плоский профиль в synthetic',
       'Интервалы помогают отличить устойчивую структуру от случайных колебаний малой выборки.',top=.84)
display(time_v9.round(3).reset_index(drop=True))
for name,df in [('Synthetic train',train),('Original',original)]:
    hours=pd.cut(df['time_of_day_hour'],bins=[-1,5,11,17,23],labels=[t[2] for t in time_blocks])
    chi2,p,_,_=chi2_contingency(pd.crosstab(hours,df[TARGET]))
    print(f'{name}: χ² = {chi2:.2f}, p = {p:.4g} (время суток × fraud)')
```

## 39 · Hidden Fingerprint: What Is Preserved Alongside Account Balance?

A large number of exact `account_balance_usd` matches does not mean that synthetic transactions were copied from original. We test a narrower hypothesis: **when the account balance matches, do other fields also match more often than they do after those fields are randomly shuffled in original?**

For seven additional features, we compare the match rate **among train rows whose balances appear in original** with a reference based on four permutations of the second field. Shuffling preserves value frequencies but breaks their association with account balance.

```python
def _shared_codes(left, right, field):
    # Build identical category codes for exact values from both sources.
    merged = pd.concat([left[field], right[field]], axis=0, ignore_index=True)
    values, _ = pd.factorize(merged, sort=False, use_na_sentinel=True)
    return values[:len(left)].astype(np.int64), values[len(left):].astype(np.int64)

def _pair_key(code_a, code_b):
    # Collision-free integer pair packing when each factor has fewer than 2^32 levels.
    a = np.asarray(code_a, dtype=np.int64).astype(np.uint32).astype(np.uint64)
    b = np.asarray(code_b, dtype=np.int64).astype(np.uint32).astype(np.uint64)
    return (a << np.uint64(32)) | b

sample_balance = train.sample(n=min(50000,len(train)), random_state=SEED+20).reset_index(drop=True)
bal_train, bal_orig = _shared_codes(sample_balance, original, 'account_balance_usd')
balance_match = np.isin(bal_train, np.unique(bal_orig))
companion_fields = CAT + ['customer_age', 'merchant_risk_score']
rng_fp = np.random.default_rng(SEED+20)
fingerprint_rows = []
for field in companion_fields:
    x_train, x_orig = _shared_codes(sample_balance, original, field)
    query = _pair_key(bal_train, x_train)
    observed = np.isin(query, np.unique(_pair_key(bal_orig, x_orig)))[balance_match].mean()
    controls = []
    for rep in range(4):
        scrambled = rng_fp.permutation(x_orig)
        controls.append(np.isin(query, np.unique(_pair_key(bal_orig, scrambled)))[balance_match].mean())
    fingerprint_rows.append({'feature':field, 'observed':observed,
                             'null_mean':float(np.mean(controls)), 'gain':observed-np.mean(controls)})
fingerprint = pd.DataFrame(fingerprint_rows).sort_values('gain')
fig, (ax1, ax2) = plt.subplots(1,2,figsize=(17,6.2), gridspec_kw={'width_ratios':[1.3,.9]})
yy = np.arange(len(fingerprint))
ax1.barh(yy-.18, fingerprint['observed']*100, height=.35, color=C['blue'], label='Exact pairs')
ax1.barh(yy+.18, fingerprint['null_mean']*100, height=.35, color=C['gold'], label='Shuffled original')
ax1.set_yticks(yy, fingerprint['feature'])
ax1.set_xlabel('Match rate among balance-matched rows, %')
ax1.set_xlim(0, max(5, fingerprint[['observed','null_mean']].to_numpy().max()*100)*1.12)
ax1.legend(frameon=False, loc='lower right', fontsize=8)
ax1.grid(axis='x', alpha=.15)
ax1.set_title('Pair Matches with Original',loc='left')
ax2.barh(yy, fingerprint['gain']*100, color=[C['mint'] if z>=0 else C['red'] for z in fingerprint['gain']])
ax2.axvline(0, color=C['muted'],lw=1)
ax2.set_yticks(yy, ['']*len(yy))
ax2.set_xlabel('Excess over control, percentage points')
ax2.set_title('Residual Association',loc='left')
ax2.grid(axis='x', alpha=.15)
finish(fig, 'Hidden Fingerprint · Does Anything Beyond Account Balance Match?',
       f'Train sample: {len(sample_balance):,}; balance matched in {balance_match.sum():,} rows; full original dataset.',top=.84)
display(fingerprint.sort_values('gain', ascending=False).round(5).reset_index(drop=True))
```

## 40 · Uneven Reuse of Original Account Balances

Exact matches in `account_balance_usd` between original and the synthetic datasets are common, but original values reappear at **very different frequencies**. In original, **19,260 distinct balances occur once**, **367 occur twice**, and **two occur three times**. Balances occurring twice in original appear in synthetic train **202 times on average**, versus about **27 times** for values that occur once. One balance that occurs twice in original appears **974 times** in train.

This unevenness is also visible in test. When original balances are ranked by how often they reappear in train, the **most frequently reused 10%** account for **53.6% of all exact train matches** and **53.3% of test matches**. Correlations between train and test reuse frequencies are **0.985 (Pearson)** and **0.963 (Spearman)**.

This indicates stable, highly uneven **reuse of individual values**, but it does not establish copying of entire original rows or transfer of original `is_fraud` labels. Rounding of balances, duplicates within original, and the shape of the original distribution can also affect exact-match counts.

```python
# Count how many times each original balance value reappears in synthetic train and test.
source_balance_counts = original['account_balance_usd'].value_counts().rename('Original count')
reuse_balance = pd.DataFrame(index=source_balance_counts.index)
reuse_balance['Original count'] = source_balance_counts
reuse_balance['Train repeats'] = train['account_balance_usd'].value_counts().reindex(reuse_balance.index).fillna(0).astype(int)
reuse_balance['Test repeats'] = test['account_balance_usd'].value_counts().reindex(reuse_balance.index).fillna(0).astype(int)
reuse_balance.index.name = 'account_balance_usd'

reuse_by_source_count = (reuse_balance.groupby('Original count')
                         .agg(original_values=('Train repeats', 'size'),
                              train_matches=('Train repeats', 'sum'),
                              test_matches=('Test repeats', 'sum'),
                              mean_train_repeats=('Train repeats', 'mean'),
                              median_train_repeats=('Train repeats', 'median')))
display(reuse_by_source_count.round(2))

ranked_balances = reuse_balance.sort_values('Train repeats', ascending=False)
top_fracs = [0.01, 0.05, 0.10, 0.25]
concentration_rows = []
for frac in top_fracs:
    selected = ranked_balances.iloc[:max(1, round(len(ranked_balances) * frac))]
    concentration_rows.append({
        'Top original values, %': 100 * frac,
        'Distinct balances': len(selected),
        'Matched train rows, %': 100 * selected['Train repeats'].sum() / reuse_balance['Train repeats'].sum(),
        'Matched test rows, %': 100 * selected['Test repeats'].sum() / reuse_balance['Test repeats'].sum()
    })
concentration_balance = pd.DataFrame(concentration_rows)
print(f'Exact original balance matches: {reuse_balance["Train repeats"].sum():,} in train; '
      f'{reuse_balance["Test repeats"].sum():,} in test.')
print(f'Train/Test repeat correlations: '
      f'Pearson {reuse_balance["Train repeats"].corr(reuse_balance["Test repeats"]):.3f}; '
      f'Spearman {reuse_balance["Train repeats"].corr(reuse_balance["Test repeats"], method="spearman"):.3f}.')
display(concentration_balance.round(2))
display(ranked_balances.head(10).reset_index().rename(columns={'account_balance_usd': 'Balance, USD'}))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6.8), gridspec_kw={'width_ratios': [0.9, 1.1]})
y = np.arange(len(concentration_balance))
ax1.barh(y - .18, concentration_balance['Matched train rows, %'],
         height=.35, color=C['blue'], label='Train')
ax1.barh(y + .18, concentration_balance['Matched test rows, %'],
         height=.35, color=C['mint'], label='Test')
ax1.set_yticks(y, [f'Top {v:g}%' for v in concentration_balance['Top original values, %']])
ax1.invert_yaxis()
ax1.set_xlim(0, 100)
ax1.set_xlabel('Share of all exact original-value matches, %')
ax1.set_title('Reuse Concentration', loc='left')
ax1.legend(frameon=False)
ax1.grid(axis='x', alpha=.14)

ax2.scatter(reuse_balance['Train repeats'] + 1, reuse_balance['Test repeats'] + 1,
            s=10, alpha=.23, color=C['blue'], edgecolors='none', rasterized=True)
ax2.set_xscale('log')
ax2.set_yscale('log')
ax2.set_xlabel('Train occurrences + 1, log scale')
ax2.set_ylabel('Test occurrences + 1, log scale')
ax2.set_title('Reuse Counts Are Stable Across Splits', loc='left')
ax2.grid(alpha=.13)
finish(fig, 'Original-Value Reuse · Concentration and Stability',
       'Top-value sets are defined by train counts and evaluated unchanged on test; all original distinct balances are included.',
       top=.90)
```

## 41 · Echo of Original Labels in Matching Balances

A matching `account_balance_usd` value alone says nothing about `is_fraud`. We ask a more specific question: is the synthetic fraud rate associated with the label of an original row that has **the same balance**?

To avoid ambiguous matches, we keep only balances that appear **exactly once** in original. Within the matching train subset, the fraud rate is approximately **1.665%** when the original label is 0 and **2.089%** when it is 1. The next cell checks whether this difference persists under negative controls. This is an association between labels, not evidence of leakage.

```python
# Keep only balances that identify exactly one row of the original dataset.
orig_unique = original.loc[~original['account_balance_usd'].duplicated(keep=False),
                           ['account_balance_usd',TARGET]].rename(columns={TARGET:'original_label'})
matched_v9 = train[['account_balance_usd',TARGET,'auth_method','merchant_category',
                    'cvv_retry_count','customer_age']].merge(orig_unique,on='account_balance_usd',how='inner')
matched_v9['original_label']=matched_v9['original_label'].astype(int)
label_echo_v9=matched_v9.groupby('original_label')[TARGET].agg(['size','sum','mean']).rename(
    columns={'size':'Строк train','sum':'Fraud train','mean':'Fraud rate'})
label_echo_v9['Fraud rate, %']=100*label_echo_v9['Fraud rate']
print(f'Уникальных балансов в original: {len(orig_unique):,}; совпавших строк train: {len(matched_v9):,}')
display(label_echo_v9[['Строк train','Fraud train','Fraud rate, %']].round(4))
fig,axes=plt.subplots(1,2,figsize=(14,5.8))
rates=label_echo_v9['Fraud rate, %'].reindex([0,1])
counts=label_echo_v9['Строк train'].reindex([0,1])
axes[0].bar(['Original label 0','Original label 1'],rates,color=[C['blue'],C['red']],width=.52)
axes[1].bar(['Original label 0','Original label 1'],counts,color=[C['blue'],C['red']],width=.52)
axes[0].set_ylabel('Fraud rate synthetic train, %')
axes[1].set_ylabel('Число совпавших строк train')
axes[1].set_yscale('log')
for ax in axes:
    ax.grid(axis='y',alpha=.15)
axes[0].set_title('Риск в train',loc='left')
axes[1].set_title('Поддержка сравнения',loc='left')
finish(fig,'Одинаковый баланс: слабое, но измеримое эхо метки original',
       'Повторные synthetic строки с одним балансом зависимы; ниже проверяем устойчивость.',top=.82)
```

## 42 · Controls for the Original Label Echo

We assess how robust the preceding difference is. As a contrast, we compare exact balance matches with artificial shifts of **1, 2, and 10 cents, and 1 dollar** in original values. We then examine the two halves of synthetic train separately, split by ID.

Differences may also occur under control shifts: balance values are discrete, and the groups linked to an original label of 1 become small. It is therefore important to show **not just the difference in fraud rates, but also the number of matched rows and fraud cases**.

```python
# Negative controls: re-link source labels using intentionally shifted original balances.
ctrl_rows=[]
orig_unique_raw=original.loc[~original['account_balance_usd'].duplicated(keep=False),
                            ['account_balance_usd',TARGET]].copy()
for shift in [0,.01,.02,.10,1.00]:
    donor=orig_unique_raw.copy()
    donor['account_balance_usd']=(donor['account_balance_usd']+shift).round(2)
    donor=donor.loc[~donor['account_balance_usd'].duplicated(keep=False)].rename(
        columns={TARGET:'donor_label'})
    joined=train[['account_balance_usd',TARGET]].merge(donor,on='account_balance_usd',how='inner')
    st=joined.groupby('donor_label')[TARGET].agg(['size','sum','mean'])
    n0=int(st.loc[0,'size']) if 0 in st.index else 0
    n1=int(st.loc[1,'size']) if 1 in st.index else 0
    f0=int(st.loc[0,'sum']) if 0 in st.index else 0
    f1=int(st.loc[1,'sum']) if 1 in st.index else 0
    dpp=100*(st.loc[1,'mean']-st.loc[0,'mean']) if (n0 and n1) else np.nan
    ctrl_rows.append({'Сдвиг original, $':shift,'Совпадений с label=0':n0,
                      'Совпадений с label=1':n1,'Fraud при label=0':f0,
                      'Fraud при label=1':f1,'Разница риска, п.п.':dpp})
controls_echo_v9=pd.DataFrame(ctrl_rows)

half_rows=[]
mid_id=train['id'].median()
for label,mask in [('Первая половина',train['id']<=mid_id),('Вторая половина',train['id']>mid_id)]:
    matched_half=train.loc[mask,['account_balance_usd',TARGET]].merge(
        orig_unique.rename(columns={'original_label':'donor_label'}),
        on='account_balance_usd',how='inner')
    st=matched_half.groupby('donor_label')[TARGET].agg(['size','sum','mean'])
    half_rows.append({'Часть train':label,'Совпадений label=1':int(st.loc[1,'size']),
                      'Fraud label=1':int(st.loc[1,'sum']),
                      'Разница риска, п.п.':100*(st.loc[1,'mean']-st.loc[0,'mean'])})
halves_echo_v9=pd.DataFrame(half_rows)
fig,ax=plt.subplots(figsize=(12,5.5))
labels=[f'+{v:g} $' for v in controls_echo_v9['Сдвиг original, $']]
values=controls_echo_v9['Разница риска, п.п.']
ax.bar(labels,values,color=[C['blue']]+[C['gold']]*4,width=.58)
ax.axhline(0,color=C['muted'],linewidth=1)
ax.set_ylabel('Разница fraud rate между donor-label 1 и 0, п.п.')
ax.set_xlabel('Искусственный сдвиг значений original')
ax.grid(axis='y',alpha=.15)
finish(fig,'Эхо исходной метки: точное совпадение и отрицательные контроли',
       'Нулевой сдвиг выделен отдельно; при других сдвигах размер совпавших групп уменьшается.',top=.83)
display(controls_echo_v9.round(4).reset_index(drop=True))
display(halves_echo_v9.round(4).reset_index(drop=True))
```

## 43 · Exact Row Overlap Between Sources

We check whether any transactions match on **all 24 input features**, excluding ID and target. This tests the simplest scenario: direct copying of an entire row.

```python
def overlap_rows(left, right, cols):
    a = pd.MultiIndex.from_frame(left[cols])
    b = pd.MultiIndex.from_frame(right[cols])
    return int(a.isin(b).sum())

full_matches = pd.DataFrame({
    'Comparison': ['Train → Original', 'Test → Original', 'Train → Test'],
    'Full-row matches': [overlap_rows(train, original, FEATURES),
                          overlap_rows(test, original, FEATURES),
                          overlap_rows(train, test, FEATURES)]
})
display(full_matches.style.hide(axis='index'))
```

## 44 · Partial Overlap: Which Dependencies Survived Generation?

We check whether **pairs of values** from original are preserved in synthetic. As a control, we randomly shuffle the second column in original: its marginal distribution stays the same, but its relationship with the first column is broken.

```python
PAIRS = [
    ('amount_usd', 'merchant_category'),
    ('amount_usd', 'auth_method'),
    ('amount_usd', 'hours_since_last_txn'),
    ('merchant_risk_score', 'merchant_category'),
    ('account_balance_usd', 'merchant_category'),
    ('velocity_score', 'auth_method')
]

rng = np.random.default_rng(SEED)
pair_rows = []
for a, b in PAIRS:
    observed = overlap_rows(train, original, [a, b]) / len(train)
    null = []
    for k in range(5):
        neg = original[[a, b]].copy()
        neg[b] = rng.permutation(neg[b].to_numpy())
        null.append(overlap_rows(train, neg, [a, b]) / len(train))
    pair_rows.append([a, b, observed, np.mean(null), np.std(null), observed - np.mean(null)])
pair_result = pd.DataFrame(pair_rows, columns=['Feature A', 'Feature B', 'Overlap', 'Control', 'Control SD', 'Difference'])
display(pair_result.sort_values('Difference', ascending=False).round(5).reset_index(drop=True))

fig, ax = plt.subplots(figsize=(12, 6))
ss = pair_result.sort_values('Difference', ascending=True)
labels = ss['Feature A'] + '\n× ' + ss['Feature B']
y = np.arange(len(ss))
ax.barh(y-.19, ss['Overlap']*100, height=.36, color=C['blue'], label='Observed')
ax.barh(y+.19, ss['Control']*100, height=.36, color=C['gold'], label='Permuted original')
ax.set_yticks(y, labels, fontsize=9)
ax.set_xlabel('Train rows with a pair found in original, %')
ax.legend(frameon=False)
ax.grid(axis='x', alpha=.13)
finish(fig, 'Were Joint Values Preserved?', 'The null model breaks associations within original; five permutations provide a reference, not an exact p-value.', top=.90)
```

## 45 · Synthetic Dependency Map: All 276 Feature Pairs

With 24 features, there are **276 unique pairs**. For each pair, we measure what share of value combinations in a fixed train subsample are also found in original. We then shuffle one original feature three times and compare the observed match rate with this **control reference**.

The key measure is **the percentage-point difference between the observed match rate and the mean rate after shuffling**. This is a map of preserved value combinations, not a Pearson correlation matrix or proof of row copying. We use up to 40,000 train rows, all of original, and a fixed random seed. Three permutations are useful for exploration but **insufficient for estimating a p-value**.

```python
from itertools import combinations

dep_sample = train.sample(n=min(40000,len(train)), random_state=SEED+22).reset_index(drop=True)
dep_codes = {field:_shared_codes(dep_sample, original, field) for field in FEATURES}
rng_dep = np.random.default_rng(SEED+22)
dep_records = []
for a,b in combinations(FEATURES,2):
    tr_a, or_a = dep_codes[a]
    tr_b, or_b = dep_codes[b]
    tr_key = _pair_key(tr_a,tr_b)
    obs = np.isin(tr_key, np.unique(_pair_key(or_a,or_b))).mean()
    controls = []
    for rep in range(3):
        shuf_b = rng_dep.permutation(or_b)
        controls.append(np.isin(tr_key, np.unique(_pair_key(or_a,shuf_b))).mean())
    dep_records.append({'a':a,'b':b,'observed':obs,'null':np.mean(controls),
                        'excess_pp':100*(obs-np.mean(controls))})
dep = pd.DataFrame(dep_records)
assert len(dep) == 276, 'Expected 276 unique pairs'
matrix = pd.DataFrame(np.nan, index=FEATURES, columns=FEATURES)
for r in dep.itertuples(index=False):
    matrix.loc[r.a,r.b] = r.excess_pp
    matrix.loc[r.b,r.a] = r.excess_pp
fig, ax = plt.subplots(figsize=(15.5,13.8))
lim = max(.1, float(np.nanquantile(np.abs(matrix.to_numpy(dtype=float)),.98)))
sns.heatmap(matrix.astype(float), ax=ax, cmap='RdBu_r', center=0, vmin=-lim, vmax=lim,
            mask=np.eye(len(FEATURES), dtype=bool), linewidths=.25,
            linecolor=C['white'], square=True, cbar_kws={'label':'Observed – control, percentage points', 'shrink':.72})
ax.tick_params(axis='x', labelrotation=75, labelsize=8)
ax.tick_params(axis='y', labelrotation=0, labelsize=8)
finish(fig, 'Synthetic Dependency Map · 276 Feature-Pair Checks',
       'Color indicates an excess or deficit of exact pairs relative to shuffled original; values are clipped at the 98th percentile.',top=.92)

top = dep.sort_values('excess_pp', ascending=False).head(14).sort_values('excess_pp')
fig, ax = plt.subplots(figsize=(13,7))
labels = (top['a']+' × '+top['b']).tolist()
yy=np.arange(len(top))
ax.barh(yy, top['excess_pp'], color=[C['mint'] if d>=0 else C['red'] for d in top['excess_pp']], height=.65)
ax.set_yticks(yy, labels, fontsize=9)
for i, val in enumerate(top['excess_pp']):
    ax.text(val + max(.04, top['excess_pp'].max()*.025), i, f'+{val:.2f} pp', va='center', fontsize=9)
ax.axvline(0, color=C['muted'], lw=1)
ax.set_xlim(min(0,top['excess_pp'].min()), max(.5,top['excess_pp'].max()*1.34))
ax.set_xlabel('Pair overlap in excess of null, percentage points')
ax.grid(axis='x', alpha=.15)
finish(fig, 'Top 14 Pairs Most Strongly Preserved vs Random Control',
       f'Exploratory ranking based on {len(dep_sample):,} train rows. Multiple testing may affect this ranking.',top=.88)
display(dep.sort_values('excess_pp',ascending=False).head(20).round(5).reset_index(drop=True))
```

## Key Findings

**Fraud is unevenly distributed.** The overall fraud rate in synthetic train is about **1.67%**, but individual purchase categories, authentication methods, and signal combinations differ in risk by orders of magnitude. A high fraud rate does not necessarily imply broad coverage of fraudulent cases or a large transaction value.

**No single warning flag explains all fraud.** Transactions with none of the six Boolean indicators account for approximately **33.1%** of fraud cases in train. Even within this group, authentication methods and repeated CVV attempts distinguish sharply different risk profiles.

**Context changes how warning signals should be read.** `billing_shipping_mismatch` occurs far more often in Online, but fraud risk among flagged transactions also depends on the channel. Flag prevalence and the probability of fraud *given that the flag is active* are not the same thing.

**Train and test are similar; original is not always similar to them.** The synthetic splits resemble each other in individual feature distributions and in the joint distributions examined. Comparisons with original, however, reveal changes in time-of-day risk, age-related risk, the `hours_since_last_txn`–`velocity_score` relationship, and other dependencies.

**The original data left measurable traces.** Exact `account_balance_usd` matches are frequent and extremely uneven. Additional checks reveal a weak link with original labels, but matching individual values **does not prove row copying or target leakage**.

### Open Questions

- What process caused the uneven reuse of original balance values?
- Why were some dependencies in original weakened while others became stronger in synthetic?
- Which feature combinations remain stable beyond the observed subgroups?

Major effects are examined alongside group sizes and control checks. Statistical uncertainty remains substantial for rare groups and for the 339 positive examples in original.

## 49 · XGBoost Experiments: From Raw Features to the Best OOF

The experiments asked a simple question: **how much can we gain by changing the way numerical values are represented?** Four feature configurations were compared using the same five-fold OOF ROC AUC protocol.

Target Encoding of the integer and decimal parts improved the result over raw features. Adding exact-value Target Encoding gave the strongest OOF score. A much larger combination with Digits, Frequency, Interactions, Profiles, and Triple TE did not improve on it, despite taking considerably longer to train.

The selected configuration reached **0.950183 OOF ROC AUC**. Its submission, `SUB_final_selected.csv`, achieved **0.95084 on the Kaggle Public Leaderboard**. The graphics below show the score progression and the trade-off between additional features, training time, and validation performance.


```python
# Results from full five-fold OOF experiments with a single XGBoost.
# Each row uses the same CV protocol. No model training is performed in this cell.
from IPython.display import HTML

cv_results = pd.DataFrame([
    ['Raw features (RAW)', 0.947273, 453.0],
    ['Integer + decimal TE', 0.949466, 525.2],
    ['Extended feature groups', 0.950059, 2809.2],
    ['Integer + decimal + exact TE', 0.950183, 594.7],
], columns=['Feature representation', 'OOF ROC AUC', 'Time (sec)'])

raw_auc = cv_results.loc[0, 'OOF ROC AUC']
cv_results['Gain vs RAW'] = cv_results['OOF ROC AUC'] - raw_auc
cv_results['Time (min)'] = cv_results['Time (sec)'] / 60
cv_results['Experiment'] = ['RAW', 'Parts TE', 'Extended features', 'Parts + exact TE']

# A matched two-panel figure: OOF gains and the time/performance trade-off.
colors = [C['muted'], C['blue'], C['gold'], C['mint']]
fig, (ax1, ax2) = plt.subplots(
    1, 2, figsize=(18, 6.2), gridspec_kw={'width_ratios': [1.35, .8]}
)
fig.subplots_adjust(wspace=.30)

positions = np.arange(len(cv_results))
ax1.barh(positions, cv_results['Gain vs RAW'], color=colors,
         height=.58, edgecolor='white', linewidth=1.1)
ax1.set_yticks(positions)
ax1.set_yticklabels(cv_results['Feature representation'], fontsize=10)
ax1.invert_yaxis()
ax1.set_xlim(-.00010, .00364)
ax1.axvline(0, color=C['muted'], linewidth=1, alpha=.65)
ax1.set_xlabel('OOF ROC AUC gain relative to RAW')
ax1.set_title('What Improved the Five-Fold OOF?', loc='left', pad=15)
ax1.grid(axis='x', alpha=.13)
ax1.set_axisbelow(True)
for i, gain in enumerate(cv_results['Gain vs RAW']):
    label = 'BASELINE' if i == 0 else f'+{gain:.6f}'
    ax1.text(max(0, gain) + .000085, i, label, va='center',
             fontsize=9, fontweight='bold', color=C['ink'])

# The winning compact TE set performed better than the much slower extended set.
for i, row in cv_results.iterrows():
    ax2.scatter(row['Time (min)'], row['OOF ROC AUC'], s=190 if i == 3 else 125,
                color=colors[i], edgecolor='white', linewidth=2, zorder=4)
    if i == 0:
        offset = (7, -17)
    elif i == 1:
        offset = (8, -18)
    elif i == 2:
        offset = (-112, 12)
    else:
        offset = (11, 10)
    ax2.annotate(row['Experiment'],
                 (row['Time (min)'], row['OOF ROC AUC']),
                 xytext=offset, textcoords='offset points',
                 fontsize=9, fontweight='bold' if i == 3 else 'normal',
                 color=C['ink'])
ax2.set_xlim(0, 54)
ax2.set_ylim(.9468, .95075)
ax2.set_xlabel('Total five-fold run time, minutes')
ax2.set_ylabel('OOF ROC AUC')
ax2.set_title('More Features Did Not Mean Better OOF', loc='left', pad=15)
ax2.grid(alpha=.13)
ax2.set_axisbelow(True)

finish(fig, 'From Raw Values to a Stronger XGBoost',
       'Four verified five-fold OOF runs; the final submission scored 0.95084 on Public LB.',
       top=.85)

# Compact scorecard matching the notebook's blue / teal visual language.
display(HTML('''''''''<div style="display:flex;flex-wrap:wrap;gap:12px;margin:10px 0 14px;font-family:Arial,sans-serif;">
<div style="flex:1 1 220px;min-width:190px;padding:16px 19px;border-radius:12px;background:#14263B;color:#FFFFFF;border:1px solid #23475F;">
<div style="font-size:11px;letter-spacing:1px;color:#9FC3D4;font-weight:700;">BEST 5-FOLD OOF ROC AUC</div>
<div style="font-size:29px;font-weight:800;margin-top:6px;color:#FFFFFF;">0.950183</div>
</div>
<div style="flex:1 1 220px;min-width:190px;padding:16px 19px;border-radius:12px;background:#F1F8F7;border:1px solid #B9DBD6;">
<div style="font-size:11px;letter-spacing:1px;color:#287B73;font-weight:700;">KAGGLE PUBLIC LEADERBOARD</div>
<div style="font-size:29px;font-weight:800;margin-top:6px;color:#146E67;">0.95084</div>
</div>
</div>'''''''''))

report_table = cv_results[['Feature representation', 'OOF ROC AUC',
                            'Gain vs RAW', 'Time (min)']].copy()

def emphasize_winner(row):
    return [
        'background-color:#E4F5EF;color:#16483E;font-weight:bold;border-top:2px solid #13A99B;'
        if row.name == 3 else ''
        for _ in row
    ]

styled = (report_table.style
          .hide(axis='index')
          .format({'OOF ROC AUC': '{:.6f}',
                   'Gain vs RAW': lambda x: '—' if abs(x) < 1e-12 else f'+{x:.6f}',
                   'Time (min)': '{:.1f}'})
          .apply(emphasize_winner, axis=1)
          .set_table_styles([
              {'selector':'th', 'props': [('background-color','#14263B'),
                                         ('color','#FFFFFF'),('padding','12px 13px'),
                                         ('text-align','left'),('font-size','12px')]},
              {'selector':'td', 'props': [('padding','11px 13px'),
                                         ('border-bottom','1px solid #DFE8EE'),
                                         ('font-size','12px')]},
              {'selector':'table','props':[('border-collapse','collapse'),
                                           ('width','100%')]},
          ])
)
display(styled)
```
